## Paso 4 — el `while` que mantiene vivo el menú

Mete el menú y el `input` del paso 3 dentro de un `while True`. Todavía sin lógica de verdad en las opciones 1, 2 y 3 — solo un `print` que diga "(aquí va a ir...)". La opción `4` sí debe terminar el ciclo de verdad, con `break`.

````python
while True:
    print("-" * 40)
    print("EL LÁPIZ FELIZ")
    print("-" * 40)
    print("1. Registrar una factura")
    print("2. Ver total acumulado del día")
    print("3. Ver cuántas facturas se han registrado")
    print("4. Salir")
    print("-" * 40)

    opcion = input("Elige una opción: ")

    if opcion == "1":
        print("(aquí va a ir registrar la factura)")
    elif opcion == "2":
        print("(aquí va a ir el total del día)")
    elif opcion == "3":
        print("(aquí va a ir el número de facturas)")
    elif opcion == "4":
        print("¡Gracias por usar el sistema! Hasta pronto.")
        break
    else:
        print("Opción no válida, intenta de nuevo.")
````

## Paso 6 — opción 1: registrar una factura con `for`

Completa la opción `1` del menú. Necesitas, en este orden:

1. Preguntar cuántos productos tiene la factura (`cantidad`, convertida a entero con `int()`)
2. Crear `total_factura` empezando en 0
3. Un `for i in range(cantidad):` que en cada vuelta pida nombre y precio, y sume el precio a `total_factura`
4. Imprimir el total de la factura
5. Sumar `total_factura` a `total_dia`
6. Sumar 1 a `facturas_hechas`

Recuerda poner `total_dia = 0` y `facturas_hechas = 0` **antes** del `while` (paso 5), no adentro de la opción 1.

````python
total_dia = 0
facturas_hechas = 0

while True:
    print("-" * 40)
    print("EL LÁPIZ FELIZ")
    print("-" * 40)
    print("1. Registrar una factura")
    print("2. Ver total acumulado del día")
    print("3. Ver cuántas facturas se han registrado")
    print("4. Salir")
    print("-" * 40)

    opcion = input("Elige una opción: ")

    if opcion == "1":
        cantidad = int(input("¿Cuántos productos tiene esta factura? "))
        total_factura = 0

        for i in range(cantidad):
            print(f"Producto {i + 1} de {cantidad}:")
            nombre = input("  Nombre del producto: ")
            precio = float(input("  Precio: "))
            total_factura = total_factura + precio

        print(f"Factura registrada. Total: ${total_factura:,.0f}")
        total_dia = total_dia + total_factura
        facturas_hechas = facturas_hechas + 1
    elif opcion == "2":
        print("(aquí va a ir el total del día)")
    elif opcion == "3":
        print("(aquí va a ir el número de facturas)")
    elif opcion == "4":
        print("¡Gracias por usar el sistema! Hasta pronto.")
        break
    else:
        print("Opción no válida, intenta de nuevo.")
````

## Paso 7 — opciones 2 y 3: mostrar lo acumulado

Completa las opciones `2` y `3`, mostrando `total_dia` y `facturas_hechas`. Esta es la versión final y completa del menú.

````python
total_dia = 0
facturas_hechas = 0

while True:
    print("-" * 40)
    print("EL LÁPIZ FELIZ")
    print("-" * 40)
    print("1. Registrar una factura")
    print("2. Ver total acumulado del día")
    print("3. Ver cuántas facturas se han registrado")
    print("4. Salir")
    print("-" * 40)

    opcion = input("Elige una opción: ")

    if opcion == "1":
        cantidad = int(input("¿Cuántos productos tiene esta factura? "))
        total_factura = 0

        for i in range(cantidad):
            print(f"Producto {i + 1} de {cantidad}:")
            nombre = input("  Nombre del producto: ")
            precio = float(input("  Precio: "))
            total_factura = total_factura + precio

        print(f"Factura registrada. Total: ${total_factura:,.0f}")
        total_dia = total_dia + total_factura
        facturas_hechas = facturas_hechas + 1
    elif opcion == "2":
        print(f"Total acumulado del día: ${total_dia:,.0f}")
    elif opcion == "3":
        print(f"Facturas registradas hoy: {facturas_hechas}")
    elif opcion == "4":
        print("¡Gracias por usar el sistema! Hasta pronto.")
        break
    else:
        print("Opción no válida, intenta de nuevo.")
````

````python
#catalogo de productos
catalogo = {
    "P-001": {"nombre": "Cuaderno cuadriculado", "precio": 8500,  "iva": 0.19},
    "E-011": {"nombre": "Caja de lapiceros x12", "precio": 15990, "iva": 0.19},
    "L-030": {"nombre": "Libro juvenil",         "precio": 45000, "iva": 0.00},
    "R-050": {"nombre": "Revista educativa",     "precio": 12000, "iva": 0.05},
}
````

````python
#Facturación

pedido = [("P-001", 3), ("L-030", 1), ("R-050", 2)]
encabezado = ("F-102", "28/08/2026", "C-02")   # tupla: no se modifica

total = 0
for codigo, cantidad in pedido:
    p = catalogo[codigo]
    subtotal = p["precio"] * cantidad
    iva = subtotal * p["iva"]
    total = total + subtotal + iva
    print(p["nombre"], cantidad, subtotal + iva)

if clientes[encabezado[2]]["tipo"] == "colegio" and total >= 200000:
    total = total * 0.9
print("TOTAL A PAGAR:", total)
````
